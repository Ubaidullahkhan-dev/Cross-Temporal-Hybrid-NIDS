```python
# ==========================================
# FINAL EVALUATION & COMPARISON
# PROJECT INITIALIZATION
# ==========================================

import os
import json
import pandas as pd
import numpy as np

print("=" * 70)
print("FINAL EVALUATION & COMPARISON")
print("=" * 70)

# Current project directory
PROJECT_PATH = os.getcwd()

print("\nProject Path:")
print(PROJECT_PATH)

# Show available items
print("\nAvailable items:")

for item in os.listdir(PROJECT_PATH):
    print("-", item)
```

```python
# ==========================================
# FIND EXPERIMENT RESULT FOLDERS
# ==========================================

print("=" * 70)
print("SEARCHING FOR EXPERIMENT RESULTS")
print("=" * 70)

experiment_folders = []

for root, dirs, files in os.walk(PROJECT_PATH):

    for folder in dirs:

        folder_lower = folder.lower()

        if (
            "exp2" in folder_lower
            or "exp3" in folder_lower
            or "exp4" in folder_lower
            or "experiment_2" in folder_lower
            or "experiment_3" in folder_lower
            or "experiment_4" in folder_lower
        ):

            full_path = os.path.join(
                root,
                folder
            )

            if full_path not in experiment_folders:
                experiment_folders.append(full_path)


print("\nFound experiment folders:")

if experiment_folders:

    for folder in experiment_folders:
        print("-", folder)

else:

    print("No experiment result folders found automatically.")
```

```python
# ==========================================
# FIND SAVED RESULT FILES
# ==========================================

print("=" * 70)
print("SEARCHING FOR SAVED RESULT FILES")
print("=" * 70)

result_files = []

extensions = (
    ".json",
    ".csv",
    ".txt",
    ".pkl",
    ".joblib"
)

for root, dirs, files in os.walk(PROJECT_PATH):

    for file in files:

        if file.lower().endswith(extensions):

            full_path = os.path.join(
                root,
                file
            )

            result_files.append(full_path)


print("\nSaved result files:")

for file in result_files:

    print("-", file)

print(
    "\nTotal result files found:",
    len(result_files)
)
```

```python
# ==========================================
# LOAD EXPERIMENT 4 FINAL RESULT
# ==========================================

exp4_files = [
    file
    for file in result_files
    if "experiment_4_final_result.json" in file.lower()
]

if len(exp4_files) == 0:

    print("Experiment 4 final result not found.")

else:

    EXP4_RESULT_FILE = exp4_files[0]

    print("Experiment 4 result found:")
    print(EXP4_RESULT_FILE)

    with open(
        EXP4_RESULT_FILE,
        "r",
        encoding="utf-8"
    ) as f:

        exp4_result = json.load(f)

    print("\nExperiment 4 contents:")

    print(
        json.dumps(
            exp4_result,
            indent=4,
            ensure_ascii=False
        )
    )
```

```python
# ==========================================
# EXPERIMENT 4 SUMMARY
# ==========================================

if "exp4_result" in globals():

    print("=" * 70)
    print("EXPERIMENT 4 SUMMARY")
    print("=" * 70)

    print(
        "Prediction:",
        exp4_result.get("prediction")
    )

    print(
        "RF Attack Probability:",
        exp4_result.get(
            "random_forest_probability"
        )
    )

    print(
        "Selected Sample:",
        exp4_result.get(
            "selected_sample_index"
        )
    )

    print(
        "\nRetrieved Knowledge:"
    )

    for item in exp4_result.get(
        "retrieved_knowledge",
        []
    ):

        print(
            "-",
            item.get("title"),
            "| Similarity:",
            round(
                item.get("Similarity", 0),
                4
            )
        )

    print("\nLLM Explanation:")
    print(
        exp4_result.get(
            "llm_explanation",
            "Not available"
        )
    )

else:

    print(
        "Experiment 4 result is not loaded."
    )
```

```python
# ==========================================
# FIND EXPERIMENT 2 RESULTS
# ==========================================

print("=" * 70)
print("SEARCHING FOR EXPERIMENT 2 RESULTS")
print("=" * 70)

exp2_files = []

for file in result_files:

    file_lower = file.lower()

    if (
        "exp2" in file_lower
        or "experiment_2" in file_lower
    ):

        exp2_files.append(file)


print("\nExperiment 2 related files:")

if exp2_files:

    for file in exp2_files:
        print("-", file)

else:

    print(
        "No Experiment 2 files found automatically."
    )
```

```python
# ==========================================
# INSPECT EXPERIMENT 2 FILES
# ==========================================

print("=" * 70)
print("EXPERIMENT 2 FILE INSPECTION")
print("=" * 70)

for file in exp2_files:

    print("\n" + "-" * 70)
    print("FILE:", file)

    try:

        if file.lower().endswith(".json"):

            with open(
                file,
                "r",
                encoding="utf-8"
            ) as f:

                data = json.load(f)

            print(
                json.dumps(
                    data,
                    indent=2,
                    ensure_ascii=False
                )[:5000]
            )

        elif file.lower().endswith(".csv"):

            df_temp = pd.read_csv(file)

            print(
                "Shape:",
                df_temp.shape
            )

            display(
                df_temp.head()
            )

        elif file.lower().endswith(".txt"):

            with open(
                file,
                "r",
                encoding="utf-8",
                errors="ignore"
            ) as f:

                text = f.read()

            print(text[:5000])

    except Exception as e:

        print(
            "Could not inspect:",
            type(e).__name__,
            e
        )
```

```python
# ==========================================
# FINAL MODEL COMPARISON
# ==========================================

print("=" * 70)
print("FINAL MODEL COMPARISON")
print("=" * 70)

comparison_data = [
    {
        "Model": "Random Forest",
        "Precision": 1.0,
        "Recall": 0.190476,
        "F1": 0.320000,
        "PR-AUC": 1.0,
        "FPR": 0.0,
        "FNR": 0.809524
    },
    {
        "Model": "Isolation Forest",
        "Precision": 0.0,
        "Recall": 0.000000,
        "F1": 0.000000,
        "PR-AUC": 1.0,
        "FPR": 0.0,
        "FNR": 1.000000
    },
    {
        "Model": "Hybrid",
        "Precision": 1.0,
        "Recall": 0.047619,
        "F1": 0.090909,
        "PR-AUC": 1.0,
        "FPR": 0.0,
        "FNR": 0.952381
    }
]

final_comparison = pd.DataFrame(
    comparison_data
)

display(final_comparison)
```

```python
# ==========================================
# BEST MODEL ANALYSIS
# ==========================================

best_f1_row = final_comparison.loc[
    final_comparison["F1"].idxmax()
]

best_recall_row = final_comparison.loc[
    final_comparison["Recall"].idxmax()
]

best_precision_row = final_comparison.loc[
    final_comparison["Precision"].idxmax()
]

print("=" * 70)
print("BEST MODEL ANALYSIS")
print("=" * 70)

print(
    "\nBest F1 Score:",
    best_f1_row["Model"],
    "->",
    best_f1_row["F1"]
)

print(
    "Best Recall:",
    best_recall_row["Model"],
    "->",
    best_recall_row["Recall"]
)

print(
    "Best Precision:",
    best_precision_row["Model"],
    "->",
    best_precision_row["Precision"]
)
```

```python
# ==========================================
# SAVE FINAL COMPARISON
# ==========================================

FINAL_RESULTS_PATH = os.path.join(
    PROJECT_PATH,
    "Final_Project_Results"
)

os.makedirs(
    FINAL_RESULTS_PATH,
    exist_ok=True
)

comparison_csv = os.path.join(
    FINAL_RESULTS_PATH,
    "final_model_comparison.csv"
)

final_comparison.to_csv(
    comparison_csv,
    index=False
)

print("=" * 70)
print("FINAL COMPARISON SAVED")
print("=" * 70)

print(comparison_csv)
```

```python
# ==========================================
# FINAL PROJECT SUMMARY
# ==========================================

summary = {
    "Experiment_2": "Completed",
    "Experiment_3_SHAP": "Completed",
    "Experiment_4_RAG_LLM": "Completed",
    "Best_F1_Model": best_f1_row["Model"],
    "Best_F1": float(best_f1_row["F1"]),
    "Best_Recall_Model": best_recall_row["Model"],
    "Best_Recall": float(best_recall_row["Recall"])
}

summary_path = os.path.join(
    FINAL_RESULTS_PATH,
    "final_project_summary.json"
)

with open(
    summary_path,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        summary,
        f,
        indent=4
    )

print("=" * 70)
print("FINAL PROJECT SUMMARY")
print("=" * 70)

print(
    json.dumps(
        summary,
        indent=4
    )
)

print(
    "\nSummary saved:",
    summary_path
)
```
