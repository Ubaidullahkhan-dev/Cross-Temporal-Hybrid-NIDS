```python
import pandas as pd


data_path = r"customer data"

# Saari Parquet files
parquet_files = []

for root, dirs, files in os.walk(data_path):
    for file in files:
        if file.endswith(".parquet"):
            parquet_files.append(os.path.join(root, file))

# Har file ka basic information
for file in parquet_files:
    df = pd.read_parquet(file)

    print("=" * 70)
    print("FILE:", os.path.basename(file))
    print("Rows:", df.shape[0])
    print("Columns:", df.shape[1])
    print("\nColumns:")
    print(df.columns.tolist())
    print("\nData Types:")
    print(df.dtypes)
    print("\nFirst 5 Rows:")
    display(df.head())
```

```python
processed_path = r"data"
for file in parquet_files:
    print("Processing",os.path.basename(file))

    original_rows=len(df)
    print("Total Rows :",original_rows)

    # 1. Clean column names
    df.columns=df.columns.str.strip()
    # 2. Clean Label values
    if "Label" in df.columns:
        df['Label']=df["Label"].astype(str).str.strip()
    # 3. Replace infinite values with NaN
    num_col=df.select_dtypes(include=np.number).columns

    df[num_col]=df[num_col].replace(
        [np.inf,-np.inf],np.nan )

     # 4. Remove duplicate rows

    dup_count=df.duplicated().sum()
    
    #df=df.drop_duplicates

    # 5. Handle missing values
    missing_before = df.isnull().sum().sum()
    

    # Remove rows containing missing values
    df = df.dropna()
    missing_after = df.isnull().sum().sum()
    

    # 6. Reset index
    df=df.reset_index(drop=True)

    #SAve Clean File
    output_file=os.path.join(
        processed_path,
        os.path.basename(file)
    )
    df.to_parquet(
        output_file,
        index=False
    )

    print("Original rows :", original_rows)
    print("Duplicates removed:", dup_count)
    print("Missing values before:", missing_before)
    print("Missing values after :", missing_after)
    print("Final rows :", len(df))
    print("Final columns :", len(df.columns))
    print("Saved to :", output_file)

print("\n" + "=" * 70)
print("DATA CLEANING COMPLETED")
print("=" * 70)
```
