## RandomForestClassifier

```python
import pandas as pd
import os
import joblib
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    classification_report,
    confusion_matrix
)
```

```python
X_train_scaled = np.load(os.path.join("scaled", "X_train_scaled.npy"))
X_val_scaled = np.load(os.path.join("scaled", "X_val_scaled.npy"))
X_test_scaled = np.load(os.path.join("scaled", "X_test_scaled.npy"))
y_train = np.load(os.path.join("scaled", "y_train.npy"))
y_val = np.load(os.path.join("scaled", "y_val.npy"))
y_test = np.load(os.path.join("scaled", "y_test.npy"))

scaler = joblib.load(os.path.join("scaled", "fitted_scaler.pkl"))
print("Data load Successfully")
```

```python
rf_model=RandomForestClassifier(n_estimators=100,max_depth=None,random_state=42,n_jobs=-1,class_weight='balanced')
rf_model.fit(X_train_scaled,y_train)
```

```python
y_val_pred = rf_model.predict(X_val_scaled)
y_val_prob = rf_model.predict_proba(X_val_scaled)[:, 1]
```

```python
print("=" * 70)
print("RANDOM FOREST - VALIDATION RESULTS")
print("=" * 70)

print("Accuracy :", accuracy_score(y_val, y_val_pred))
print("Precision:", precision_score(y_val, y_val_pred, zero_division=0))
print("Recall   :", recall_score(y_val, y_val_pred, zero_division=0))
print("F1 Score :", f1_score(y_val, y_val_pred, zero_division=0))
print("ROC-AUC  :", roc_auc_score(y_val, y_val_prob))

print("\nClassification Report:")
print(classification_report(y_val, y_val_pred, zero_division=0))

print("\nConfusion Matrix:")
print(confusion_matrix(y_val, y_val_pred))
```

```python
# Test

y_test_pred = rf_model.predict(X_test_scaled)
y_test_prob = rf_model.predict_proba(X_test_scaled)[:, 1]

print("\n" + "=" * 70)
print("RANDOM FOREST - TEST RESULTS")
print("=" * 70)

print("Accuracy :", accuracy_score(y_test, y_test_pred))
print("Precision:", precision_score(y_test, y_test_pred, zero_division=0))
print("Recall   :", recall_score(y_test, y_test_pred, zero_division=0))
print("F1 Score :", f1_score(y_test, y_test_pred, zero_division=0))
print("ROC-AUC  :", roc_auc_score(y_test, y_test_prob))

print("\nClassification Report:")
print(classification_report(y_test, y_test_pred, zero_division=0))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_test_pred))
```

##  IsolationForest

```python
from sklearn.ensemble import IsolationForest
from sklearn.metrics import classification_report, confusion_matrix, roc_auc_score

# Isolation Forest

iso_model = IsolationForest(
    n_estimators=100,
    contamination="auto",
    random_state=42,
    n_jobs=-1
)

iso_model.fit(X_train_scaled)
```

```python
# Validation

y_val_iso = iso_model.predict(X_val_scaled)

# Isolation Forest:
# +1 = Normal
# -1 = Anomaly
y_val_anomaly = (y_val_iso == -1).astype(int)

print("=" * 70)
print("ISOLATION FOREST - VALIDATION RESULTS")
print("=" * 70)

print("\nClassification Report:")
print(classification_report(
    y_val,
    y_val_anomaly,
    zero_division=0
))

print("\nConfusion Matrix:")
print(confusion_matrix(y_val, y_val_anomaly))


```

```python

# Test

y_test_iso = iso_model.predict(X_test_scaled)

y_test_anomaly = (y_test_iso == -1).astype(int)

print("\n" + "=" * 70)
print("ISOLATION FOREST - TEST RESULTS")
print("=" * 70)

print("\nClassification Report:")
print(classification_report(
    y_test,
    y_test_anomaly,
    zero_division=0
))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_test_anomaly))



# Anomaly Scores


val_anomaly_score = -iso_model.score_samples(X_val_scaled)
test_anomaly_score = -iso_model.score_samples(X_test_scaled)

print("\n" + "=" * 70)
print("ANOMALY SCORES")
print("=" * 70)

print("Validation score range:",
      val_anomaly_score.min(),
      "to",
      val_anomaly_score.max())

print("Test score range:",
      test_anomaly_score.min(),
      "to",
      test_anomaly_score.max())
```

## HYBRID RISK SCORE

```python
# NORMALIZE ISOLATION FOREST SCORES


val_iso_min = val_anomaly_score.min()
val_iso_max = val_anomaly_score.max()

# Min-Max Scaling Formula: (Score - Min) / (Max - Min)
val_iso_norm = (
    (val_anomaly_score - val_iso_min)
    / (val_iso_max - val_iso_min + 1e-10)
)

test_iso_norm = (
    (test_anomaly_score - val_iso_min)
    / (val_iso_max - val_iso_min + 1e-10)
)

# Keep test values within 0-1
test_iso_norm = np.clip(test_iso_norm, 0, 1)


# HYBRID RISK SCORE

alpha = 0.5

val_hybrid_risk = (
    alpha * y_val_prob
    +
    (1 - alpha) * val_iso_norm
)

test_hybrid_risk = (
    alpha * y_test_prob
    +
    (1 - alpha) * test_iso_norm
)


# CHECK RESULTS

print("=" * 70)
print("HYBRID RISK SCORE COMPLETED")
print("=" * 70)

print("\nValidation Hybrid Risk:")
print("Min :", val_hybrid_risk.min())
print("Max :", val_hybrid_risk.max())
print("Mean:", val_hybrid_risk.mean())

print("\nTest Hybrid Risk:")
print("Min :", test_hybrid_risk.min())
print("Max :", test_hybrid_risk.max())
print("Mean:", test_hybrid_risk.mean())

print("\nFirst 10 Test Hybrid Scores:")
print(test_hybrid_risk[:10])
```
